```python
import sys
import os
import glob
from io import StringIO

# 1. CLEANUP: Hapus cache dan file sisa hari sebelumnya agar tidak konflik
files_to_clean = ['macro_v2_cache.csv', 'full_execution_log.txt', 'MacroDashboardV2.png']
for f in files_to_clean:
    if os.path.exists(f): os.remove(f)

# Hapus semua potongan gambar lama
for p in glob.glob("*.png"):
    if "horizontal_part" in p or "part_" in p:
        os.remove(p)

# 2. Setup logging capture
old_stdout = sys.stdout
new_stdout = StringIO()
sys.stdout = new_stdout

try:
    print("Starting Global Macro Strategy V2 (Fresh Daily Run)...")
    # Menjalankan script utama
    exec(open("drive/MyDrive/GlobalMacroStrategyV2.py").read())
except Exception as e:
    print(f"\n❌ ERROR DURING EXECUTION: {str(e)}")
finally:
    # 3. Restore stdout and save to file
    output = new_stdout.getvalue()
    sys.stdout = old_stdout

    with open("full_execution_log.txt", "w") as f:
        f.write(output)

    print("✅ Execution finished. Log saved to full_execution_log.txt")
    print("Preview of output:")
    print(output[:500] + "...")
```

```python
import requests
import os
import json
from PIL import Image

def send_macro_report_v3(image_path, log_path, token, chat_id):
    url_media = f'https://api.telegram.org/bot{token}/sendMediaGroup'
    url_doc = f'https://api.telegram.org/bot{token}/sendDocument'

    if not os.path.exists(image_path):
        print('❌ Dashboard image not found.')
        return

    img = Image.open(image_path)
    w, h = img.size
    section_h = h // 4

    files = {}
    media = []

    for i in range(4):
        top = i * section_h
        bottom = (i + 1) * section_h if i < 3 else h
        part_name = f'horizontal_part_{i+1}.png'
        img.crop((0, top, w, bottom)).save(part_name)

        files[f'photo{i}'] = open(part_name, 'rb')
        media.append({
            'type': 'photo',
            'media': f'attach://photo{i}',
            'caption': f'— Macro Report Section {i+1} —' if i == 0 else ''
        })

    # Send images
    requests.post(url_media, data={'chat_id': chat_id, 'media': json.dumps(media)}, files=files)

    # Send the FULL LOG file
    if os.path.exists(log_path):
        with open(log_path, 'rb') as f:
            requests.post(url_doc, data={'chat_id': chat_id, 'caption': '📄 Full Strategy Execution Log'}, files={'document': f})
        print(f'✅ Full log ({os.path.getsize(log_path)} bytes) sent to Telegram.')
    else:
        print('❌ Log file not found.')

# Configuration
TOKEN = '8610930703:AAGS6k4udQnxcik1FzXwDOCYVa6e4Bpcwp4'
CHAT_ID = '7656599370'
IMG_PATH = '/content/MacroDashboardV2.png'
LOG_PATH = 'full_execution_log.txt'

send_macro_report_v3(IMG_PATH, LOG_PATH, TOKEN, CHAT_ID)
```
